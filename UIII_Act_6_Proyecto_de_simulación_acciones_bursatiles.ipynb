### Acciones bursátiles
### ASESOR ESTRATEGICO DE INVERSIONES

```python
# Simulación de IA para Análisis de Acciones Bursátiles
print("--- SISTEMA IA: ASESOR ESTRATÉGICO DE INVERSIONES ---")

# Variable de control para el bucle interactivo
continuar = "si"

# Bucle while para permitir múltiples consultas [1]
while continuar == "si":
    print("\n--- Nueva Evaluación de Activo ---")

    # Entrada de datos interactiva [2, 3]
    empresa = input("Ingrese el nombre o Ticker de la acción: ")
    precio_actual = float(input(f"Precio actual de {empresa}: "))
    variacion_diaria = float(input("Variación porcentual hoy (ej: 2.5 o -1.2): "))
    volumen_operado = int(input("Volumen de transacciones (en unidades): "))
    tendencia_mercado = input("¿La tendencia del mercado es Alcista o Bajista?: ").lower()

    # Uso de operadores aritméticos para calcular un 'Índice de Riesgo' [4, 5]
    # El riesgo aumenta si la variación es muy brusca respecto al precio
    indice_riesgo = (abs(variacion_diaria) * 100) / precio_actual

    print(f"\nAnalizando datos de {empresa}...")

    # Lógica de decisión de la IA utilizando operadores relacionales y lógicos [6-8]
    if tendencia_mercado == "alcista" and variacion_diaria > 3.0 and volumen_operado > 10000:
        # Condición para Compra Fuerte
        print("Veredicto IA: ¡COMPRA FUERTE! El impulso y el volumen validan la tendencia.")
    elif tendencia_mercado == "bajista" or variacion_diaria < -5.0:
        # Condición para Venta por riesgo de caída [8]
        if indice_riesgo > 5.0:
            print("Veredicto IA: VENTA INMEDIATA. El índice de riesgo es demasiado alto para este activo.")
        else:
            print("Veredicto IA: VENTA SUGERIDA. El mercado no favorece este movimiento.")
    elif variacion_diaria >= -1.0 and variacion_diaria <= 1.0:
        # Condición de estabilidad
        print("Veredicto IA: MANTENER (HOLD). La acción muestra consolidación lateral.")
    else:
        # Caso por defecto para situaciones de incertidumbre [9]
        print("Veredicto IA: NEUTRAL. Los datos son insuficientes para una predicción clara.")

    # Mensaje adicional basado en el cálculo de riesgo
    if indice_riesgo < 0.5:
        print("Nota de la IA: Este activo se considera de baja volatilidad actualmente.")

    # Instrucción para decidir si repetir el ciclo [12]
    continuar = input("\n¿Desea analizar otra acción? (si/no): ").lower()

print("\nGracias por utilizar el Asesor IA. ¡Buenas inversiones!")
```
